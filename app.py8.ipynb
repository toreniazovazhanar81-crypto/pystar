```python
print("🌟 PYSTAR")
print("==========================")

name = input("👤 Оқушының аты: ")

score = 0

print()
print("🎮 Қош келдің,", name, "!")
print("⭐ Бастапқы StarCoin:", score)

print()
print("📚 МИССИЯЛАР")
print("1️⃣ Математика")
print("2️⃣ Құпия код")
print("3️⃣ Робот энергиясы")

choice = int(input("🎯 Миссияны таңда: "))

print()

if choice == 1:
    answer = int(input("🧠 7 × 6 = ? "))

    if answer == 42:
        print("🎉 Дұрыс!")
        score = score + 10
    else:
        print("❌ Қате жауап.")

elif choice == 2:
    code = int(input("🔐 Құпия кодты енгіз: "))

    if code == 1234:
        print("🎉 Код дұрыс!")
        score = score + 20
    else:
        print("❌ Код қате.")

elif choice == 3:
    energy = int(input("⚡ Роботқа қанша энергия бересің? "))
    score = score + energy
    print("🤖 Роботқа", energy, "энергия берілді!")

else:
    print("❌ Мұндай миссия жоқ.")

if score >= 41:
    level = 3
elif score >= 21:
    level = 2
else:
    level = 1

print()
print("==========================")
print("🏆 НӘТИЖЕ")
print("👤 Оқушы:", name)
print("⭐ StarCoin:", score)
print("🏆 Level:", level)
print("==========================")

if level == 3:
    print("🔥 Сен — Python шеберісің!")
elif level == 2:
    print("👏 Өте жақсы! Алға!")
else:
    print("🚀 Жаттығуды жалғастыр!")
```

```python
print("🌟 PYSTAR — 11-МИССИЯ")
print("==========================")

name = input("👤 Оқушының аты: ")
score = 0

print()
print("🎮 Сәлем,", name, "!")
print("🚀 3 тапсырманы орында!")

# 1-тапсырма
print()
print("🧠 1-ТАПСЫРМА")
answer = int(input("5 + 7 = ? "))

if answer == 12:
    print("✅ Дұрыс! +10 StarCoin")
    score = score + 10
else:
    print("❌ Қате.")

# 2-тапсырма
print()
print("🔐 2-ТАПСЫРМА")
code = int(input("Құпия код: "))

if code == 1234:
    print("✅ Дұрыс! +10 StarCoin")
    score = score + 10
else:
    print("❌ Қате.")

# 3-тапсырма
print()
print("🤖 3-ТАПСЫРМА")
energy = int(input("Роботқа қанша энергия бересің? "))

if energy >= 20:
    print("✅ Робот іске қосылды! +10 StarCoin")
    score = score + 10
else:
    print("❌ Энергия жеткіліксіз.")

# Level
if score >= 30:
    level = 3
elif score >= 20:
    level = 2
else:
    level = 1

print()
print("==========================")
print("🏆 ОЙЫН НӘТИЖЕСІ")
print("👤 Оқушы:", name)
print("⭐ StarCoin:", score)
print("🏆 Level:", level)
print("==========================")
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
    background: linear-gradient(135deg, #667eea, #764ba2);
    color: white;
    padding: 30px;
    border-radius: 25px;
    text-align: center;
    font-family: Arial;
    max-width: 500px;
    margin: auto;
">

<h1>🌟 PyStar</h1>
<h2>🏆 Level таңда</h2>

<button onclick="startLevel(1)"
style="padding:12px 25px; margin:5px; border-radius:12px;">
🟢 Level 1
</button>

<button onclick="startLevel(2)"
style="padding:12px 25px; margin:5px; border-radius:12px;">
🟡 Level 2
</button>

<button onclick="startLevel(3)"
style="padding:12px 25px; margin:5px; border-radius:12px;">
🔴 Level 3
</button>

<div id="mission" style="margin-top:25px; font-size:20px;"></div>

</div>

<script>
function startLevel(level) {

    if (level == 1) {
        document.getElementById("mission").innerHTML =
        "🟢 Level 1<br><br>🧠 5 + 3 = ?";
    }

    if (level == 2) {
        document.getElementById("mission").innerHTML =
        "🟡 Level 2<br><br>🧠 12 × 4 = ?";
    }

    if (level == 3) {
        document.getElementById("mission").innerHTML =
        "🔴 Level 3<br><br>🧠 25 × 8 = ?";
    }
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
    background: linear-gradient(135deg, #667eea, #764ba2);
    color: white;
    padding: 30px;
    border-radius: 25px;
    text-align: center;
    font-family: Arial;
    max-width: 500px;
    margin: auto;
">

<h1>🌟 PyStar</h1>
<h2>🎮 Миссия таңда</h2>

<button onclick="startLevel(1)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🟢 Level 1
</button>

<button onclick="startLevel(2)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🟡 Level 2
</button>

<button onclick="startLevel(3)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🔴 Level 3
</button>

<div id="game" style="margin-top:25px;"></div>

<script>
let correctAnswer = 0;

function startLevel(level) {

    let question = "";

    if (level == 1) {
        question = "🧠 5 + 3 = ?";
        correctAnswer = 8;
    }

    if (level == 2) {
        question = "🧠 12 × 4 = ?";
        correctAnswer = 48;
    }

    if (level == 3) {
        question = "🧠 25 × 8 = ?";
        correctAnswer = 200;
    }

    document.getElementById("game").innerHTML = `
        <h2>${question}</h2>

        <input id="answer"
        type="number"
        placeholder="Жауабың"
        style="
            padding:12px;
            border-radius:10px;
            border:none;
            font-size:18px;
            width:150px;
        ">

        <br><br>

        <button onclick="checkAnswer()"
        style="
            padding:12px 25px;
            border:none;
            border-radius:12px;
            font-size:17px;
            font-weight:bold;
        ">
        ✅ ТЕКСЕРУ
        </button>

        <p id="result" style="font-size:20px;"></p>
    `;
}

function checkAnswer() {

    let answer = Number(document.getElementById("answer").value);

    if (answer === correctAnswer) {
        document.getElementById("result").innerHTML =
        "🎉 ДҰРЫС! ⭐ +10 StarCoin";
    } else {
        document.getElementById("result").innerHTML =
        "❌ Қате жауап. Қайта байқап көр!";
    }
}
</script>

</div>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
    background: linear-gradient(135deg, #667eea, #764ba2);
    color: white;
    padding: 30px;
    border-radius: 25px;
    text-align: center;
    font-family: Arial;
    max-width: 500px;
    margin: auto;
">

<h1>🌟 PyStar</h1>

<h2>⭐ StarCoin: <span id="score">0</span></h2>

<p style="font-size:20px;">🧠 6 + 4 = ?</p>

<input id="answer"
type="number"
placeholder="Жауабың"
style="
    padding:12px;
    border-radius:10px;
    border:none;
    font-size:18px;
    width:150px;
">

<br><br>

<button onclick="checkAnswer()"
style="
    padding:12px 25px;
    border:none;
    border-radius:12px;
    font-size:17px;
    font-weight:bold;
">
✅ ТЕКСЕРУ
</button>

<p id="result" style="font-size:20px;"></p>

</div>

<script>
let score = 0;
let answered = false;

function checkAnswer() {

    if (answered) {
        return;
    }

    let answer = Number(document.getElementById("answer").value);

    if (answer === 10) {

        score = score + 10;

        document.getElementById("score").innerHTML = score;

        document.getElementById("result").innerHTML =
        "🎉 ДҰРЫС! ⭐ +10 StarCoin";

        answered = true;

    } else {

        document.getElementById("result").innerHTML =
        "❌ Қате! Қайта байқап көр.";
    }
}
</script>

"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
    background: linear-gradient(135deg, #667eea, #764ba2);
    color: white;
    padding: 30px;
    border-radius: 25px;
    text-align: center;
    font-family: Arial;
    max-width: 600px;
    margin: auto;
">

<h1>🌟 PyStar</h1>
<h2>🎓 23-Миссия</h2>

<p style="font-size:20px;">
Сыныбыңды таңда:
</p>

<button onclick="startGame(7)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🟢 7-сынып
</button>

<button onclick="startGame(8)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🟡 8-сынып
</button>

<button onclick="startGame(9)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🔴 9-сынып
</button>

<button onclick="startGame(10)"
style="padding:12px 20px; margin:5px; border-radius:12px;">
🟣 10-сынып
</button>

<h2>⭐ StarCoin: <span id="score">0</span></h2>

<div id="game" style="margin-top:25px;"></div>

</div>

<script>

let correctAnswer = 0;
let score = 0;

function startGame(grade) {

    let question = "";

    if (grade == 7) {

        question = "🧠 2x + 7 = 19<br><br>x = ?";
        correctAnswer = 6;

    }

    else if (grade == 8) {

        question = "🧠 x² - 5x + 6 = 0<br><br>Үлкен түбірді тап:";
        correctAnswer = 3;

    }

    else if (grade == 9) {

        question = "🧠 2x² - 7x + 3 = 0<br><br>Үлкен түбірді тап:";
        correctAnswer = 3;

    }

    else if (grade == 10) {

        question = "🧠 2ˣ = 32<br><br>x = ?";
        correctAnswer = 5;

    }

    document.getElementById("game").innerHTML = `

        <h3>${grade}-сынып тапсырмасы</h3>

        <p style="font-size:23px;">
        ${question}
        </p>

        <input id="answer"
        type="number"
        placeholder="Жауабың"
        style="
            padding:12px;
            border-radius:10px;
            border:none;
            font-size:18px;
            width:160px;
        ">

        <br><br>

        <button onclick="checkAnswer()"
        style="
            padding:12px 25px;
            border:none;
            border-radius:12px;
            font-size:17px;
            font-weight:bold;
        ">
        ✅ ТЕКСЕРУ
        </button>

        <p id="result" style="font-size:20px;"></p>
    `;
}

function checkAnswer() {

    let answer = Number(document.getElementById("answer").value);

    if (answer === correctAnswer) {

        score += 10;

        document.getElementById("score").innerHTML = score;

        document.getElementById("result").innerHTML =
        "🎉 ДҰРЫС! ⭐ +10 StarCoin";

    } else {

        document.getElementById("result").innerHTML =
        "❌ Қате. Қайта ойлан! 💡";
    }
}

</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#f3e8ff;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:600px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 7-Миссия — РОБОТ ЭНЕРГИЯСЫ</h2>

<p style="font-size:20px;">
⚡ Роботтың бастапқы энергиясын енгіз:
</p>

<input type="text" id="energyInput"
placeholder="Мысалы: 5"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="runRobot()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 РОБОТТЫ ІСКЕ ҚОС
</button>

<div id="output" style="
margin-top:25px;
font-size:20px;
line-height:1.7;">
</div>

</div>

<script>
function runRobot() {

    let energy = parseInt(
        document.getElementById("energyInput").value
    );

    if (isNaN(energy) || energy < 1) {
        document.getElementById("output").innerHTML =
        "⚠️ 1 немесе одан үлкен сан енгізіңіз!";
        return;
    }

    let result = "🤖 Робот іске қосылды!<br><br>";

    while (energy > 0) {
        result += "⚡ Энергия: " + energy + "<br>";
        energy = energy - 1;
    }

    result += "<br>🔋 Энергия: 0<br>";
    result += "🛑 Робот тоқтады!<br><br>";
    result += "⭐ +10 StarCoin";

    document.getElementById("output").innerHTML = result;
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#fce7f3;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:600px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 42-Миссия — LEVEL UP</h2>

<p style="font-size:19px;">
⚡ Энергияны енгіз:
</p>

<input type="text" id="energy42"
placeholder="Мысалы: 10"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="start42()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 LEVEL UP
</button>

<div id="result42"
style="margin-top:25px;font-size:20px;line-height:1.8;">
</div>

</div>

<script>
function start42() {

    let energy = parseInt(
        document.getElementById("energy42").value
    );

    if (isNaN(energy) || energy < 1) {
        document.getElementById("result42").innerHTML =
        "⚠️ 1 немесе одан үлкен сан енгізіңіз!";
        return;
    }

    let level = 1;
    let steps = 0;

    while (energy > 0) {

        energy = energy - 1;
        steps = steps + 1;

        if (steps % 3 == 0) {
            level = level + 1;
        }
    }

    document.getElementById("result42").innerHTML =
    "🛑 Энергия: 0" +
    "<br>🔢 Қадам саны: " + steps +
    "<br>🏆 Робот деңгейі: " + level;
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#e0e7ff;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:600px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 43-Миссия — ҚАДАМДАРДЫ ТАЛДА</h2>

<p style="font-size:19px;">
🔢 Қадам санын енгіз:
</p>

<input type="text" id="steps43"
placeholder="Мысалы: 8"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="start43()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 ТАЛДАУ
</button>

<div id="result43"
style="margin-top:25px;font-size:20px;line-height:1.8;">
</div>

</div>

<script>
function start43() {

    let steps = parseInt(
        document.getElementById("steps43").value
    );

    if (isNaN(steps) || steps < 1) {
        document.getElementById("result43").innerHTML =
        "⚠️ 1 немесе одан үлкен сан енгізіңіз!";
        return;
    }

    let i = 1;
    let even = 0;
    let odd = 0;

    while (i <= steps) {

        if (i % 2 == 0) {
            even = even + 1;
        } else {
            odd = odd + 1;
        }

        i = i + 1;
    }

    document.getElementById("result43").innerHTML =
    "🔢 Барлық қадам: " + steps +
    "<br>🟢 Жұп қадамдар: " + even +
    "<br>🟡 Тақ қадамдар: " + odd;
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#fef3c7;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:600px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 44-Миссия — САНДЫ ТАЛДА</h2>

<p style="font-size:19px;">
🔢 Бүтін сан енгіз:
</p>

<input type="text" id="number44"
placeholder="Мысалы: 5832"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="start44()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 ЕСЕПТЕУ
</button>

<div id="result44"
style="margin-top:25px;font-size:20px;line-height:1.8;">
</div>

</div>

<script>
function start44() {

    let number = parseInt(
        document.getElementById("number44").value
    );

    if (isNaN(number) || number < 1) {
        document.getElementById("result44").innerHTML =
        "⚠️ Оң бүтін сан енгізіңіз!";
        return;
    }

    let original = number;
    let sum = 0;

    while (number > 0) {

        let digit = number % 10;
        sum = sum + digit;
        number = Math.floor(number / 10);
    }

    document.getElementById("result44").innerHTML =
    "🔢 Сан: " + original +
    "<br>➕ Цифрлар қосындысы: " + sum +
    "<br>⭐ +10 StarCoin";
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#e0f2fe;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:600px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 51-Миссия — ЖАЙ САН</h2>

<p style="font-size:19px;">
🔢 Сан енгіз:
</p>

<input type="text" id="number51"
placeholder="Мысалы: 17"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="start51()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 ТЕКСЕРУ
</button>

<div id="result51"
style="margin-top:25px;font-size:20px;line-height:1.8;">
</div>

</div>

<script>
function start51() {

    let number = parseInt(
        document.getElementById("number51").value
    );

    if (isNaN(number) || number < 2) {
        document.getElementById("result51").innerHTML =
        "⚠️ 2 немесе одан үлкен сан енгізіңіз!";
        return;
    }

    let divisor = 2;
    let isPrime = true;

    while (divisor < number) {

        if (number % divisor == 0) {
            isPrime = false;
            break;
        }

        divisor = divisor + 1;
    }

    if (isPrime) {
        document.getElementById("result51").innerHTML =
        "✅ " + number + " — ЖАЙ САН!<br>" +
        "🏆 +30 StarCoin";
    } else {
        document.getElementById("result51").innerHTML =
        "❌ " + number + " — ЖАЙ САН ЕМЕС.";
    }
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#ede9fe;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:650px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 52-Миссия — ЖАЙ САНДАР</h2>

<p style="font-size:19px;">
🔢 Шекті санды енгіз:
</p>

<input type="text" id="number52"
placeholder="Мысалы: 20"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="start52()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 ЖАЙ САНДАРДЫ ТАП
</button>

<div id="result52"
style="margin-top:25px;font-size:20px;line-height:1.8;">
</div>

</div>

<script>
function start52() {

    let limit = parseInt(
        document.getElementById("number52").value
    );

    if (isNaN(limit) || limit < 2) {
        document.getElementById("result52").innerHTML =
        "⚠️ 2 немесе одан үлкен сан енгізіңіз!";
        return;
    }

    let number = 2;
    let primes = [];

    while (number <= limit) {

        let divisor = 2;
        let isPrime = true;

        while (divisor < number) {

            if (number % divisor == 0) {
                isPrime = false;
                break;
            }

            divisor = divisor + 1;
        }

        if (isPrime) {
            primes.push(number);
        }

        number = number + 1;
    }

    document.getElementById("result52").innerHTML =
    "🔢 " + limit + " дейінгі жай сандар:<br>" +
    "🟢 " + primes.join(", ") +
    "<br><br>🏆 +40 StarCoin";
}
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="
background:#fef3c7;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
max-width:650px;
margin:auto;">

<h1>🌟 PyStar</h1>
<h2>🤖 58-Миссия — ФИБОНАЧЧИ ТАЛДАУ</h2>

<p style="font-size:19px;">
🔢 Фибоначчидің қанша санын қарастырамыз?
</p>

<input type="text" id="number58"
placeholder="Мысалы: 10"
style="padding:12px;font-size:18px;width:180px;">

<br><br>

<button onclick="start58()"
style="padding:12px 25px;font-size:18px;border-radius:10px;">
🚀 ТАЛДАУ
</button>

<div id="result58"
style="margin-top:25px;font-size:20px;line-height:1.8;">
</div>

</div>

<script>
function start58() {

    let count = parseInt(
        document.getElementById("number58").value
    );

    if (isNaN(count) || count < 2) {
        document.getElementById("result58").innerHTML =
        "⚠️ 2 немесе одан үлкен сан енгізіңіз!";
        return;
    }

    let a = 0;
    let b = 1;
    let i = 0;
    let maxNumber = 0;

    while (i < count) {

        if (a > maxNumber) {
            maxNumber = a;
        }

        let next = a + b;
        a = b;
        b = next;

        i = i + 1;
    }

    document.getElementById("result58").innerHTML =
    "🔢 Сандар саны: " + count +
    "<br>🏆 Ең үлкен сан: " + maxNumber +
    "<br>⭐ +60 StarCoin";
}
</script>
"""))
```

```python
# ⭐ PyStar — 100-Миссия
# Қорытынды жоба

def analyze_scores(scores):
    average = sum(scores) / len(scores)
    highest = max(scores)
    lowest = min(scores)

    if average >= 80:
        level = "ЖОҒАРЫ"
    elif average >= 60:
        level = "ОРТАША"
    else:
        level = "ТӨМЕН"

    return average, highest, lowest, level


scores = [76, 94, 81, 67, 89, 92]

average, highest, lowest, level = analyze_scores(scores)

print("⭐ PyStar — Оқушы нәтижесі")
print("-------------------------")
print("Ұпайлар:", scores)
print("Орташа ұпай:", average)
print("Ең жоғары:", highest)
print("Ең төмен:", lowest)
print("Деңгей:", level)
```

```python
# ⭐ PyStar — 109-Миссия

def complete_mission(score):
    if score >= 80:
        starcoin = 10
        status = "🎉 Миссия орындалды!"
    elif score >= 50:
        starcoin = 5
        status = "👍 Миссия орындалды!"
    else:
        starcoin = 0
        status = "📚 Қайта дайындал!"

    return starcoin, status


scores = [92, 76, 45]

for score in scores:
    starcoin, status = complete_mission(score)

    print("Ұпай:", score)
    print("Нәтиже:", status)
    print("⭐ StarCoin:", starcoin)
    print("-------------------")
```

```python
# ⭐ PyStar — 110-Миссия
# 🎮 Шағын ойын

def play_mission(answer, correct_answer, starcoin):
    if answer == correct_answer:
        starcoin += 10
        return "🎉 Дұрыс жауап!", starcoin
    else:
        return "❌ Қате жауап!", starcoin


starcoin = 0

print("⭐ PyStar — Python Challenge")
print("----------------------------")
print("Сұрақ: Python-да тізімнің соңына элемент қосатын команда қандай?")
print("1) remove()")
print("2) append()")
print("3) sort()")

answer = "append()"

message, starcoin = play_mission(answer, "append()", starcoin)

print()
print(message)
print("⭐ StarCoin:", starcoin)
```

```python
# ⭐ PyStar — 111-Миссия

def check_answer(answer, correct_answer, starcoin):
    if answer == correct_answer:
        starcoin += 10
        print("🎉 Дұрыс!")
    else:
        print("❌ Қате!")

    return starcoin


starcoin = 0

missions = [
    ["Python-да экранға шығару командасы?", "print()"],
    ["Тізімге элемент қосу командасы?", "append()"],
    ["Тізімді сұрыптау командасы?", "sort()"]
]

for mission in missions:
    question = mission[0]
    correct_answer = mission[1]

    print("\n🎯", question)
    print("Дұрыс жауап:", correct_answer)

    answer = correct_answer

    starcoin = check_answer(answer, correct_answer, starcoin)

print("\n🏆 Ойын аяқталды!")
print("⭐ Жалпы StarCoin:", starcoin)
```

```python
# ⭐ PyStar — 137-Миссия

def calculate_starcoin(score):
    if score >= 90:
        return 30
    elif score >= 75:
        return 20
    elif score >= 50:
        return 10
    else:
        return 0


def get_level(starcoin):
    if starcoin >= 30:
        return "MASTER"
    elif starcoin >= 20:
        return "PRO"
    elif starcoin >= 10:
        return "ADVANCED"
    else:
        return "BEGINNER"


players = [
    {"name": "Аружан", "grade": 7, "score": 95},
    {"name": "Нұрбек", "grade": 8, "score": 87},
    {"name": "Мадина", "grade": 9, "score": 76},
    {"name": "Данияр", "grade": 10, "score": 43}
]

while True:
    print("\n⭐ PYSTAR")
    print("----------------")
    print("1 — Ойыншыларды көру")
    print("2 — Рейтингті көру")
    print("3 — Профильді көру")
    print("4 — Ойыннан шығу")

    choice = input("Таңдауыңыз: ")

    if choice == "1":
        print("\n👥 ОЙЫНШЫЛАР")

        for player in players:
            starcoin = calculate_starcoin(player["score"])
            level = get_level(starcoin)

            print(
                player["name"],
                "| Ұпай:", player["score"],
                "| ⭐", starcoin,
                "|", level
            )

    elif choice == "2":
        players.sort(
            key=lambda player: player["score"],
            reverse=True
        )

        print("\n🏆 РЕЙТИНГ")

        for number, player in enumerate(players, start=1):
            print(
                number,
                "орын:",
                player["name"],
                "-",
                player["score"],
                "ұпай"
            )

    elif choice == "3":
        name = input("\nОйыншының атын енгізіңіз: ")

        found = False

        for player in players:
            if player["name"].lower() == name.lower():

                starcoin = calculate_starcoin(player["score"])
                level = get_level(starcoin)

                print("\n👤 ОЙЫНШЫ ПРОФИЛІ")
                print("----------------")
                print("Аты:", player["name"])
                print("Сыныбы:", player["grade"])
                print("Ұпайы:", player["score"])
                print("⭐ StarCoin:", starcoin)
                print("🏆 Деңгейі:", level)

                found = True
                break

        if not found:
            print("\n❌ Ойыншы табылмады.")

    elif choice == "4":
        print("\n👋 PyStar-дан шықтыңыз!")
        break

    else:
        print("\n❌ Қате таңдау!")
```

```python
def get_level(correct_count, total):
    percentage = correct_count / total * 100

    if percentage == 100:
        return "MASTER"
    elif percentage >= 75:
        return "PRO"
    elif percentage >= 50:
        return "ADVANCED"
    else:
        return "BEGINNER"


correct_count = 3
total = 3

percentage = correct_count / total * 100
level = get_level(correct_count, total)

print("⭐ PYSTAR — НӘТИЖЕ")
print("====================")
print("Дұрыс жауап:", correct_count)
print("Барлық тапсырма:", total)
print("Нәтиже:", percentage, "%")
print("🏆 Деңгей:", level)
```

```python
def calculate_result(correct, total):
    percentage = correct / total * 100

    if percentage == 100:
        level = "MASTER"
    elif percentage >= 75:
        level = "PRO"
    elif percentage >= 50:
        level = "ADVANCED"
    else:
        level = "BEGINNER"

    return percentage, level


questions = [
    {"question": "2 + 2 = ?", "answer": "4"},
    {"question": "Python-да мәтін шығару командасы?", "answer": "print"},
    {"question": "Python-да мәлімет енгізу командасы?", "answer": "input"},
    {"question": "Шарт тексеру сөзі?", "answer": "if"},
    {"question": "3 * 3 = ?", "answer": "9"}
]

correct = 0

print("⭐ PYSTAR — ТЕСТ")
print("====================")

for number, question in enumerate(questions, start=1):

    print("\n🎯", number, "-сұрақ")
    print("❓", question["question"])

    answer = input("Жауабыңыз: ")

    if answer.strip().lower() == question["answer"].lower():
        print("✅ Дұрыс!")
        correct += 1
    else:
        print("❌ Қате!")
        print("Дұрыс жауап:", question["answer"])


total = len(questions)

percentage, level = calculate_result(correct, total)

print("\n🏆 ҚОРЫТЫНДЫ")
print("====================")
print("Дұрыс жауап:", correct)
print("Барлық сұрақ:", total)
print("Нәтиже:", round(percentage, 1), "%")
print("🏆 Деңгей:", level)
```

```python
def get_result(correct, total):
    percentage = correct / total * 100

    if percentage >= 90:
        starcoin = 30
        level = "MASTER"
    elif percentage >= 75:
        starcoin = 20
        level = "PRO"
    elif percentage >= 50:
        starcoin = 10
        level = "ADVANCED"
    else:
        starcoin = 0
        level = "BEGINNER"

    return percentage, starcoin, level


students = [
    {"name": "Аружан", "correct": 9, "total": 10},
    {"name": "Нұрбек", "correct": 8, "total": 10},
    {"name": "Мадина", "correct": 6, "total": 10},
    {"name": "Данияр", "correct": 4, "total": 10}
]

print("⭐ PYSTAR — ОҚУШЫЛАР НӘТИЖЕСІ")
print("==============================")

for student in students:

    percentage, starcoin, level = get_result(
        student["correct"],
        student["total"]
    )

    print("\n👤", student["name"])
    print("Дұрыс жауап:", student["correct"])
    print("Нәтиже:", round(percentage, 1), "%")
    print("⭐ StarCoin:", starcoin)
    print("🏆 Деңгей:", level)
```

```python
def get_result(correct, total):
    percentage = correct / total * 100

    if percentage >= 90:
        starcoin = 30
        level = "MASTER"
    elif percentage >= 75:
        starcoin = 20
        level = "PRO"
    elif percentage >= 50:
        starcoin = 10
        level = "ADVANCED"
    else:
        starcoin = 0
        level = "BEGINNER"

    return percentage, starcoin, level


students = [
    {"name": "Аружан", "correct": 9, "total": 10},
    {"name": "Нұрбек", "correct": 8, "total": 10},
    {"name": "Мадина", "correct": 6, "total": 10},
    {"name": "Данияр", "correct": 4, "total": 10}
]


# Нәтижені есептеу
for student in students:
    percentage, starcoin, level = get_result(
        student["correct"],
        student["total"]
    )

    student["percentage"] = percentage
    student["starcoin"] = starcoin
    student["level"] = level


# Нәтиже бойынша сұрыптау
students.sort(
    key=lambda student: student["percentage"],
    reverse=True
)


print("🏆 PYSTAR — РЕЙТИНГ")
print("========================")

for number, student in enumerate(students, start=1):
    print(
        number, "орын:",
        student["name"],
        "|", round(student["percentage"], 1), "%",
        "| ⭐", student["starcoin"],
        "|", student["level"]
    )
```

```python
students = [
    {"name": "Аружан", "percentage": 95, "starcoin": 30, "level": "MASTER"},
    {"name": "Нұрбек", "percentage": 82, "starcoin": 20, "level": "PRO"},
    {"name": "Дана", "percentage": 76, "starcoin": 20, "level": "PRO"},
    {"name": "Айбек", "percentage": 64, "starcoin": 10, "level": "ADVANCED"},
    {"name": "Мадина", "percentage": 48, "starcoin": 0, "level": "BEGINNER"}
]

# Жалпы StarCoin
total_starcoin = sum(student["starcoin"] for student in students)

# Орташа нәтиже
average_percentage = sum(
    student["percentage"] for student in students
) / len(students)

# Ең үздік оқушы
best_student = max(
    students,
    key=lambda student: student["percentage"]
)

print("======================================")
print("       ⭐ PYSTAR — СЫНЫП СТАТИСТИКАСЫ")
print("======================================")

print("👥 Оқушы саны:", len(students))
print("📊 Орташа нәтиже:", round(average_percentage, 1), "%")
print("⭐ Жалпы StarCoin:", total_starcoin)

print("--------------------------------------")
print("🏆 ЕҢ ҮЗДІК ОҚУШЫ")
print("Аты:", best_student["name"])
print("Нәтижесі:", best_student["percentage"], "%")
print("Деңгейі:", best_student["level"])
print("======================================")
```

```python
def check_progress(old_score, new_score):
    difference = new_score - old_score

    if difference > 0:
        return "📈 Өсті", difference
    elif difference < 0:
        return "📉 Төмендеді", abs(difference)
    else:
        return "➡️ Өзгермеді", 0


students = [
    {"name": "Аружан", "old": 82, "new": 95},
    {"name": "Нұрбек", "old": 88, "new": 82},
    {"name": "Дана", "old": 64, "new": 76},
    {"name": "Мадина", "old": 48, "new": 48}
]

print("======================================")
print("      ⭐ PYSTAR — ОҚУШЫ ПРОГРЕСІ")
print("======================================")

for student in students:
    status, difference = check_progress(
        student["old"],
        student["new"]
    )

    print("👤", student["name"])
    print("Бұрынғы нәтиже:", student["old"], "%")
    print("Қазіргі нәтиже:", student["new"], "%")
    print("Нәтиже:", status)

    if difference > 0:
        print("Өзгеріс:", difference, "%")

    print("--------------------------------------")
```

```python
def get_progress_reward(old_score, new_score):
    difference = new_score - old_score

    if difference >= 20:
        return 20
    elif difference >= 10:
        return 10
    elif difference > 0:
        return 5
    else:
        return 0


students = [
    {"name": "Аружан", "old": 70, "new": 95},
    {"name": "Нұрбек", "old": 75, "new": 85},
    {"name": "Дана", "old": 60, "new": 65},
    {"name": "Мадина", "old": 50, "new": 45}
]

print("======================================")
print("    ⭐ PYSTAR — ПРОГРЕСС СЫЙЛЫҒЫ")
print("======================================")

for student in students:
    reward = get_progress_reward(
        student["old"],
        student["new"]
    )

    difference = student["new"] - student["old"]

    print("👤", student["name"])
    print("📊 Бұрын:", student["old"], "%")
    print("📈 Қазір:", student["new"], "%")
    print("⬆️ Өзгеріс:", difference, "%")
    print("⭐ Қосымша StarCoin:", reward)
    print("--------------------------------------")
```

```python
def get_basic_starcoin(score):
    if score >= 90:
        return 30
    elif score >= 75:
        return 20
    elif score >= 50:
        return 10
    else:
        return 0


def get_progress_reward(old_score, new_score):
    difference = new_score - old_score

    if difference >= 20:
        return 20
    elif difference >= 10:
        return 10
    elif difference > 0:
        return 5
    else:
        return 0


students = [
    {"name": "Аружан", "old": 70, "new": 95},
    {"name": "Нұрбек", "old": 75, "new": 85},
    {"name": "Дана", "old": 60, "new": 65},
    {"name": "Мадина", "old": 50, "new": 45}
]

# Жалпы StarCoin есептеу
for student in students:
    basic = get_basic_starcoin(student["new"])
    bonus = get_progress_reward(
        student["old"],
        student["new"]
    )

    student["starcoin"] = basic + bonus


# StarCoin бойынша сұрыптау
students.sort(
    key=lambda student: student["starcoin"],
    reverse=True
)


print("======================================")
print("       ⭐ PYSTAR — STARCOIN РЕЙТИНГІ")
print("======================================")

for rank, student in enumerate(students, start=1):
    print(
        rank,
        "орын —",
        student["name"],
        "| ⭐ StarCoin:",
        student["starcoin"]
    )

print("======================================")
```

```python
def find_student(students, name):
    for student in students:
        if student["name"].lower() == name.lower():
            return student
    return None


def show_profile(student):
    print("======================================")
    print("          ⭐ PYSTAR ПРОФИЛІ")
    print("======================================")
    print("👤 Оқушы:", student["name"])
    print("📊 Нәтиже:", student["percentage"], "%")
    print("📈 Прогресс:", student["progress"], "%")
    print("⭐ StarCoin:", student["starcoin"])
    print("🎮 Деңгей:", student["level"])
    print("🔥 Миссия:", student["missions"])
    print("🏆 Жетістіктер:", len(student["achievements"]))
    print("======================================")


students = [
    {
        "name": "Аружан",
        "percentage": 95,
        "progress": 25,
        "starcoin": 50,
        "level": "MASTER",
        "missions": 12,
        "achievements": ["StarCoin Master", "Үздік нәтиже"]
    },
    {
        "name": "Нұрбек",
        "percentage": 82,
        "progress": 10,
        "starcoin": 30,
        "level": "PRO",
        "missions": 8,
        "achievements": ["Жақсы нәтиже"]
    },
    {
        "name": "Дана",
        "percentage": 65,
        "progress": 5,
        "starcoin": 15,
        "level": "ADVANCED",
        "missions": 6,
        "achievements": ["Алғашқы қадам"]
    }
]

name = input("🔎 Оқушының атын енгізіңіз: ")

student = find_student(students, name)

if student is not None:
    show_profile(student)
else:
    print("❌ Мұндай оқушы табылмады.")
```

```python
def get_next_level(starcoin):
    if starcoin >= 50:
        return "👑 MAX LEVEL", 0
    elif starcoin >= 30:
        return "💎 MASTER", 50 - starcoin
    elif starcoin >= 20:
        return "🚀 PRO", 30 - starcoin
    elif starcoin >= 10:
        return "⚡ ADVANCED", 20 - starcoin
    else:
        return "🌱 BEGINNER", 10 - starcoin


students = [
    {"name": "Аружан", "starcoin": 45},
    {"name": "Дана", "starcoin": 28},
    {"name": "Нұрбек", "starcoin": 17},
    {"name": "Мадина", "starcoin": 6}
]

print("======================================")
print("      ⭐ PYSTAR — КЕЛЕСІ ДЕҢГЕЙ")
print("======================================")

for student in students:
    level, needed = get_next_level(student["starcoin"])

    print("👤", student["name"])
    print("⭐ StarCoin:", student["starcoin"])
    print("🎮 Деңгей:", level)

    if needed > 0:
        print("🎯 Келесі деңгейге:", needed, "StarCoin қалды")
    else:
        print("🏆 Барлық деңгей ашылды!")

    print("--------------------------------------")
```

```python
def get_progress(starcoin):
    if starcoin >= 50:
        return 100
    elif starcoin >= 30:
        return int((starcoin / 50) * 100)
    elif starcoin >= 20:
        return int((starcoin / 30) * 100)
    elif starcoin >= 10:
        return int((starcoin / 20) * 100)
    else:
        return int((starcoin / 10) * 100)


students = [
    {"name": "Аружан", "starcoin": 45},
    {"name": "Дана", "starcoin": 28},
    {"name": "Нұрбек", "starcoin": 17},
    {"name": "Мадина", "starcoin": 6}
]

print("======================================")
print("       ⭐ PYSTAR — ПРОГРЕСС")
print("======================================")

for student in students:
    progress = get_progress(student["starcoin"])

    print("👤", student["name"])
    print("⭐ StarCoin:", student["starcoin"])
    print("📈 Прогресс:", progress, "%")

    # Қарапайым визуалды progress bar
    bar = "█" * (progress // 10)
    empty = "░" * (10 - progress // 10)

    print("[" + bar + empty + "]")
    print("--------------------------------------")

print("======================================")
```

```python
def get_next_level_info(xp):
    if xp >= 500:
        return "👑 LEGEND", 0
    elif xp >= 400:
        return "🔥 MASTER", 500 - xp
    elif xp >= 300:
        return "🚀 PRO", 400 - xp
    elif xp >= 150:
        return "⭐ ADVANCED", 300 - xp
    else:
        return "🌱 BEGINNER", 150 - xp


students = [
    {"name": "Аружан", "total_xp": 475},
    {"name": "Мадина", "total_xp": 425},
    {"name": "Дана", "total_xp": 350},
    {"name": "Нұрбек", "total_xp": 250},
    {"name": "Айбек", "total_xp": 100}
]

print("======================================")
print("      ⭐ PYSTAR — КЕЛЕСІ ДЕҢГЕЙ")
print("======================================")

for student in students:
    level, needed_xp = get_next_level_info(student["total_xp"])

    print("👤", student["name"])
    print("✨ XP:", student["total_xp"])
    print("🎮 Деңгей:", level)

    if needed_xp > 0:
        print("🎯 Келесі деңгейге:", needed_xp, "XP қалды")
    else:
        print("🏆 Ең жоғары деңгейге жетті!")

    print("--------------------------------------")

print("======================================")
```

```python
def get_xp_achievement(xp):
    achievements = []

    if xp >= 500:
        achievements.append("👑 XP Legend")

    if xp >= 300:
        achievements.append("🔥 XP Master")

    if xp >= 150:
        achievements.append("⭐ XP Hunter")

    if xp < 150:
        achievements.append("🌱 XP Beginner")

    return achievements


students = [
    {"name": "Аружан", "total_xp": 475},
    {"name": "Мадина", "total_xp": 425},
    {"name": "Дана", "total_xp": 350},
    {"name": "Нұрбек", "total_xp": 250},
    {"name": "Айбек", "total_xp": 100}
]

# Жетістіктерді есептеу
for student in students:
    student["achievements"] = get_xp_achievement(student["total_xp"])

# 🏆 Рейтинг:
# 1) жетістік саны көп болса — жоғары
# 2) жетістік саны бірдей болса — XP көп болса жоғары
students.sort(
    key=lambda s: (len(s["achievements"]), s["total_xp"]),
    reverse=True
)

print("======================================")
print("     🏆 PYSTAR — ЖЕТІСТІК РЕЙТИНГІ")
print("======================================")

for rank, student in enumerate(students, start=1):

    print(f"{rank}-орын 🏅 {student['name']}")
    print("   ✨ XP:", student["total_xp"])
    print("   🏆 Жетістік саны:", len(student["achievements"]))

    for achievement in student["achievements"]:
        print("      ", achievement)

    print("--------------------------------------")

print("======================================")
```

```python
questions = [
    {
        "question": "2 + 2 = ?",
        "answer": "4"
    },
    {
        "question": "Python-да мәтін шығару командасы?",
        "answer": "print"
    },
    {
        "question": "Python-да қайталану циклі қалай жазылады?",
        "answer": "for"
    }
]

starcoin = 0
correct = 0

print("======================================")
print("        ⭐ PYSTAR — ТЕСТ")
print("======================================")

for number, question in enumerate(questions, start=1):

    print("\n🎯", number, "-сұрақ")
    print("❓", question["question"])

    while True:
        answer = input("Жауабыңыз: ").strip().lower()

        if answer == "":
            print("⚠️ Жауап бос болмауы керек!")
            continue

        break

    if answer == question["answer"]:
        print("✅ Дұрыс!")
        correct += 1
        starcoin += 10
    else:
        print("❌ Қате!")
        print("Дұрыс жауап:", question["answer"])

print("\n======================================")
print("          🏆 НӘТИЖЕ")
print("======================================")
print("✅ Дұрыс жауап:", correct)
print("📝 Барлық тапсырма:", len(questions))
print("⭐ StarCoin:", starcoin)
print("======================================")30
```

```python
questions = [
    {"question": "2 + 2 = ?", "answer": "4"},
    {"question": "Python-да мәтін шығару командасы?", "answer": "print"},
    {"question": "Python-да қайталану циклі?", "answer": "for"}
]

correct = 0
starcoin = 0

print("======================================")
print("        ⭐ PYSTAR — ТЕСТ")
print("======================================")

for number, question in enumerate(questions, start=1):
    print("\n🎯", number, "-сұрақ")
    print("❓", question["question"])

    answer = input("Жауабыңыз: ").strip().lower()

    if answer == question["answer"]:
        print("✅ Дұрыс!")
        correct += 1
        starcoin += 10
    else:
        print("❌ Қате!")
        print("Дұрыс жауап:", question["answer"])

total = len(questions)
percentage = int((correct / total) * 100)

print("\n======================================")
print("             🏆 НӘТИЖЕ")
print("======================================")
print("✅ Дұрыс жауап:", correct)
print("📝 Барлық тапсырма:", total)
print("📊 Нәтиже:", percentage, "%")
print("⭐ StarCoin:", starcoin)
print("======================================")
```

```python
def get_level(percentage):
    if percentage >= 90:
        return "🌟 MASTER"
    elif percentage >= 75:
        return "🚀 PRO"
    elif percentage >= 50:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


percentage = int(input("📊 Нәтижеңізді енгізіңіз (0-100): "))

if 0 <= percentage <= 100:
    level = get_level(percentage)

    print("\n======================================")
    print("       ⭐ PYSTAR — НӘТИЖЕ")
    print("======================================")
    print("📊 Нәтиже:", percentage, "%")
    print("🎮 Деңгей:", level)
    print("======================================")
else:
    print("❌ 0 мен 100 аралығындағы сан енгізіңіз!")
```

```python

def create_profile(name, percentage):

    if percentage >= 90:
        level = "🌟 MASTER"
        starcoin = 30
    elif percentage >= 75:
        level = "🚀 PRO"
        starcoin = 20
    elif percentage >= 50:
        level = "⭐ ADVANCED"
        starcoin = 10
    else:
        level = "🌱 BEGINNER"
        starcoin = 0

    return {
        "name": name,
        "percentage": percentage,
        "level": level,
        "starcoin": starcoin
    }


students = [
    create_profile("Аружан", 95),
    create_profile("Мадина", 82),
    create_profile("Дана", 68),
    create_profile("Нұрбек", 45),
    create_profile("Айбек", 91)
]

students.sort(
    key=lambda student: student["percentage"],
    reverse=True
)

name = input("🔎 Оқушының атын енгізіңіз: ")

found = False

for rank, student in enumerate(students, start=1):

    if student["name"].lower() == name.lower():

        print("\n======================================")
        print("        🔎 PYSTAR — ОҚУШЫ")
        print("======================================")
        print("👤 Аты:", student["name"])
        print("🏆 Рейтингі:", rank, "-орын")
        print("📊 Нәтиже:", student["percentage"], "%")
        print("🎮 Деңгей:", student["level"])
        print("⭐ StarCoin:", student["starcoin"])
        print("======================================")

        found = True
        break

if not found:
    print("\n❌ Мұндай оқушы табылмады.")
```

```python
def get_xp_level(xp):
    if xp >= 500:
        return "👑 LEGEND"
    elif xp >= 400:
        return "🔥 MASTER"
    elif xp >= 300:
        return "🚀 PRO"
    elif xp >= 150:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


def get_next_level(xp):
    if xp >= 500:
        return "👑 LEGEND", 0
    elif xp >= 400:
        return "👑 LEGEND", 500 - xp
    elif xp >= 300:
        return "🔥 MASTER", 400 - xp
    elif xp >= 150:
        return "🚀 PRO", 300 - xp
    else:
        return "⭐ ADVANCED", 150 - xp


name = input("👤 Оқушының аты: ")
xp = int(input("✨ XP енгізіңіз: "))

if xp >= 0:

    current_level = get_xp_level(xp)
    next_level, needed_xp = get_next_level(xp)

    print("\n======================================")
    print("       🚀 PYSTAR — XP ПРОФИЛЬ")
    print("======================================")
    print("👤 Оқушы:", name)
    print("✨ XP:", xp)
    print("🎮 Қазіргі деңгей:", current_level)

    if needed_xp == 0:
        print("🏆 Барлық деңгейлер ашылды!")
    else:
        print("🔓 Келесі деңгей:", next_level)
        print("📈 Қажет XP:", needed_xp)

    print("======================================")

else:
    print("❌ XP теріс сан болмауы керек!")
```

```python
def get_progress(xp):
    if xp >= 500:
        return 100
    elif xp >= 400:
        return (xp - 400) / 100 * 100
    elif xp >= 300:
        return (xp - 300) / 100 * 100
    elif xp >= 150:
        return (xp - 150) / 150 * 100
    else:
        return xp / 150 * 100


def get_level(xp):
    if xp >= 500:
        return "👑 LEGEND"
    elif xp >= 400:
        return "🔥 MASTER"
    elif xp >= 300:
        return "🚀 PRO"
    elif xp >= 150:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


name = input("👤 Оқушының аты: ")
xp = int(input("✨ XP енгізіңіз: "))

if xp >= 0:

    level = get_level(xp)
    progress = get_progress(xp)

    # 10 бөліктен тұратын прогресс жолағы
    filled = int(progress / 10)
    bar = "🟩" * filled + "⬜" * (10 - filled)

    print("\n======================================")
    print("       📈 PYSTAR — XP ПРОГРЕСС")
    print("======================================")
    print("👤 Оқушы:", name)
    print("✨ XP:", xp)
    print("🎮 Деңгей:", level)
    print("📊 Прогресс:", round(progress), "%")
    print("[" + bar + "]")
    print("======================================")

else:
    print("❌ XP теріс сан болмауы керек!")
```

```python
def get_achievements(xp):
    achievements = []

    if xp >= 100:
        achievements.append("🥉 100 XP — Алғашқы қадам")

    if xp >= 250:
        achievements.append("🥈 250 XP — Жақсы нәтиже")

    if xp >= 400:
        achievements.append("🥇 400 XP — Нағыз шебер")

    if xp >= 500:
        achievements.append("👑 500 XP — XP LEGEND")

    return achievements


name = input("👤 Оқушының аты: ")
xp = int(input("✨ XP енгізіңіз: "))

if xp >= 0:

    achievements = get_achievements(xp)

    print("\n======================================")
    print("      🏆 PYSTAR — ЖЕТІСТІКТЕР")
    print("======================================")
    print("👤 Оқушы:", name)
    print("✨ XP:", xp)

    if len(achievements) == 0:
        print("🌱 Әзірге жетістік жоқ.")
    else:
        print("🏆 Жетістіктер:")

        for achievement in achievements:
            print("  ", achievement)

    print("📊 Жетістік саны:", len(achievements))
    print("======================================")

else:
    print("❌ XP теріс сан болмауы керек!")
```

```python
def get_achievements(xp):
    achievements = []

    if xp >= 100:
        achievements.append("🥉 100 XP — Алғашқы қадам")

    if xp >= 250:
        achievements.append("🥈 250 XP — Жақсы нәтиже")

    if xp >= 400:
        achievements.append("🥇 400 XP — Нағыз шебер")

    if xp >= 500:
        achievements.append("👑 500 XP — XP LEGEND")

    return achievements


def get_achievement_starcoin(achievements):
    return len(achievements) * 10


name = input("👤 Оқушының аты: ")
xp = int(input("✨ XP енгізіңіз: "))

if xp >= 0:

    achievements = get_achievements(xp)
    bonus_starcoin = get_achievement_starcoin(achievements)

    print("\n======================================")
    print("       🏆 PYSTAR — БОНУС")
    print("======================================")
    print("👤 Оқушы:", name)
    print("✨ XP:", xp)
    print("🏆 Жетістіктер саны:", len(achievements))
    print("⭐ Achievement бонусы:", bonus_starcoin, "StarCoin")

    for achievement in achievements:
        print("  ", achievement)

    print("======================================")

else:
    print("❌ XP теріс сан болмауы керек!")
```

```python
def get_starcoin(xp):
    if xp >= 500:
        return 50
    elif xp >= 400:
        return 40
    elif xp >= 300:
        return 30
    elif xp >= 150:
        return 20
    else:
        return 10


def get_achievements(xp):
    achievements = []

    if xp >= 100:
        achievements.append("🥉 100 XP — Алғашқы қадам")

    if xp >= 250:
        achievements.append("🥈 250 XP — Жақсы нәтиже")

    if xp >= 400:
        achievements.append("🥇 400 XP — Нағыз шебер")

    if xp >= 500:
        achievements.append("👑 500 XP — XP LEGEND")

    return achievements


name = input("👤 Оқушының аты: ")
xp = int(input("✨ XP енгізіңіз: "))

if xp >= 0:

    basic_starcoin = get_starcoin(xp)

    achievements = get_achievements(xp)

    achievement_bonus = len(achievements) * 10

    total_starcoin = basic_starcoin + achievement_bonus

    print("\n======================================")
    print("       ⭐ PYSTAR — ЖАЛПЫ ҰПАЙ")
    print("======================================")
    print("👤 Оқушы:", name)
    print("✨ XP:", xp)
    print("⭐ Негізгі StarCoin:", basic_starcoin)
    print("🏆 Жетістік саны:", len(achievements))
    print("🎁 Achievement бонусы:", achievement_bonus)
    print("--------------------------------------")
    print("💰 ЖАЛПЫ STARCOIN:", total_starcoin)
    print("======================================")

else:
    print("❌ XP теріс сан болмауы керек!")
```

```python
def get_level(xp):
    if xp >= 500:
        return "👑 LEGEND"
    elif xp >= 400:
        return "🔥 MASTER"
    elif xp >= 300:
        return "🚀 PRO"
    elif xp >= 150:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


def get_achievements(xp):
    achievements = []

    if xp >= 100:
        achievements.append("🥉 Алғашқы қадам")

    if xp >= 250:
        achievements.append("🥈 Жақсы нәтиже")

    if xp >= 400:
        achievements.append("🥇 Нағыз шебер")

    if xp >= 500:
        achievements.append("👑 XP LEGEND")

    return achievements


def get_progress(xp):
    if xp >= 500:
        return 100
    elif xp >= 400:
        return (xp - 400) / 100 * 100
    elif xp >= 300:
        return (xp - 300) / 100 * 100
    elif xp >= 150:
        return (xp - 150) / 150 * 100
    else:
        return xp / 150 * 100


def create_profile(name, xp):

    level = get_level(xp)
    achievements = get_achievements(xp)
    achievement_bonus = len(achievements) * 10
    progress = get_progress(xp)

    profile = {
        "name": name,
        "xp": xp,
        "level": level,
        "achievements": achievements,
        "achievement_bonus": achievement_bonus,
        "progress": progress
    }

    return profile


print("======================================")
print("      👤 PYSTAR — ТОЛЫҚ ПРОФИЛЬ")
print("======================================")

name = input("👤 Оқушының аты: ")
xp = int(input("✨ XP енгізіңіз: "))

if xp >= 0:

    profile = create_profile(name, xp)

    filled = int(profile["progress"] / 10)
    bar = "🟩" * filled + "⬜" * (10 - filled)

    print("\n👤 Оқушы:", profile["name"])
    print("✨ XP:", profile["xp"])
    print("🎮 Деңгей:", profile["level"])
    print("📈 Прогресс:", round(profile["progress"]), "%")
    print("[" + bar + "]")
    print("🏆 Жетістік саны:", len(profile["achievements"]))
    print("🎁 Achievement бонусы:",
          profile["achievement_bonus"], "StarCoin")

    print("\n🏆 Жетістіктер:")
    for achievement in profile["achievements"]:
        print("  ", achievement)

    print("======================================")

else:
    print("❌ XP теріс сан болмауы керек!")
```

```python
def get_xp(score):
    if score >= 90:
        return 100
    elif score >= 75:
        return 75
    elif score >= 50:
        return 50
    else:
        return 25


def get_level(xp):
    if xp >= 500:
        return "👑 LEGEND"
    elif xp >= 400:
        return "🔥 MASTER"
    elif xp >= 300:
        return "🚀 PRO"
    elif xp >= 150:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


def get_starcoin(score):
    if score >= 90:
        return 30
    elif score >= 75:
        return 20
    elif score >= 50:
        return 10
    else:
        return 0


def get_achievements(xp):
    achievements = []

    if xp >= 100:
        achievements.append("🥉 Алғашқы қадам")

    if xp >= 250:
        achievements.append("🥈 Жақсы нәтиже")

    if xp >= 400:
        achievements.append("🥇 Нағыз шебер")

    if xp >= 500:
        achievements.append("👑 XP LEGEND")

    return achievements


def get_recommendation(score):
    if score >= 90:
        return "🌟 Күрделі Python тапсырмаларын орындаңыз!"
    elif score >= 75:
        return "🚀 Python білімін тереңдетіңіз!"
    elif score >= 50:
        return "📚 Тақырыптарды қайталаңыз!"
    else:
        return "💪 Python негіздерін қайта оқыңыз!"


print("======================================")
print("       🤖 PYSTAR — AI ҰСТАЗ")
print("======================================")

name = input("👤 Оқушының аты: ")

while True:

    try:
        score = int(input("📊 Тест нәтижесі (0-100): "))

        if 0 <= score <= 100:
            break
        else:
            print("⚠️ 0-100 аралығында сан енгізіңіз!")

    except ValueError:
        print("❌ Тек сан енгізіңіз!")


# Негізгі есептеулер
xp = get_xp(score)
starcoin = get_starcoin(score)
level = get_level(xp)
achievements = get_achievements(xp)
recommendation = get_recommendation(score)


# Оқушы профилі
profile = {
    "name": name,
    "score": score,
    "xp": xp,
    "starcoin": starcoin,
    "level": level,
    "achievements": achievements,
    "recommendation": recommendation
}


# Нәтиже
print("\n======================================")
print("        🏆 PYSTAR — ПРОФИЛЬ")
print("======================================")

print("👤 Оқушы:", profile["name"])
print("📊 Нәтиже:", profile["score"], "%")
print("✨ XP:", profile["xp"])
print("⭐ StarCoin:", profile["starcoin"])
print("🎮 Деңгей:", profile["level"])

print("\n🏆 Жетістіктер:")

if len(profile["achievements"]) == 0:
    print("   🌱 Әзірге жетістік жоқ.")
else:
    for achievement in profile["achievements"]:
        print("  ", achievement)

print("\n🤖 AI ұсынысы:")
print("   ", profile["recommendation"])

print("======================================")
print("🎉 PYSTAR — МИССИЯ 200 АЯҚТАЛДЫ!")
print("🚀 СЕН PYTHON ЖОЛЫНДА ҮЛКЕН ҚАДАМ ЖАСАДЫҢ!")
print("======================================")
```

```python
# 🐍 PyStar — 213-Миссия

class Player:
    def __init__(self, name):
        self.name = name
        self.xp = 0
        self.starcoin = 0
        self.missions_completed = 0

    def get_level(self):
        if self.xp >= 500:
            return "👑 LEGEND"
        elif self.xp >= 400:
            return "🔥 MASTER"
        elif self.xp >= 300:
            return "🚀 PRO"
        elif self.xp >= 150:
            return "⭐ ADVANCED"
        else:
            return "🌱 BEGINNER"

    def complete_mission(self, xp_bonus, starcoin_bonus):
        self.xp += xp_bonus
        self.starcoin += starcoin_bonus
        self.missions_completed += 1

    def show_profile(self):
        print("================================")
        print("🤖 PyStar PLAYER")
        print("================================")
        print("👤 Оқушы:", self.name)
        print("🎯 Миссия:", self.missions_completed)
        print("⚡ XP:", self.xp)
        print("🪙 StarCoin:", self.starcoin)
        print("🏆 Деңгей:", self.get_level())
        print("================================")


player = Player("Аружан")

player.complete_mission(50, 10)
player.show_profile()

print("\n🎯 Келесі миссия орындалды!")

player.complete_mission(100, 20)
player.show_profile()
```

```python
# 🐍 PyStar — 214-Миссия

class Player:
    def __init__(self, name):
        self.name = name
        self.xp = 0

    def get_level(self):
        if self.xp >= 500:
            return "👑 LEGEND"
        elif self.xp >= 400:
            return "🔥 MASTER"
        elif self.xp >= 300:
            return "🚀 PRO"
        elif self.xp >= 150:
            return "⭐ ADVANCED"
        else:
            return "🌱 BEGINNER"

    def get_next_level(self):
        if self.xp >= 500:
            return "🏆 Максималды деңгей"
        elif self.xp >= 400:
            return f"👑 LEGEND-ке дейін: {500 - self.xp} XP"
        elif self.xp >= 300:
            return f"🔥 MASTER-ге дейін: {400 - self.xp} XP"
        elif self.xp >= 150:
            return f"🚀 PRO-ға дейін: {300 - self.xp} XP"
        else:
            return f"⭐ ADVANCED-ке дейін: {150 - self.xp} XP"


player = Player("Аружан")

player.xp = 225

print("👤 Оқушы:", player.name)
print("⚡ XP:", player.xp)
print("🏆 Деңгей:", player.get_level())
print("🎯", player.get_next_level())
```

```python
# 🐍 PyStar — 215-Миссия

class Player:
    def __init__(self, name, xp):
        self.name = name
        self.xp = xp

    def get_progress(self):
        if self.xp >= 500:
            return 100

        if self.xp >= 400:
            progress = ((self.xp - 400) / 100) * 100
        elif self.xp >= 300:
            progress = ((self.xp - 300) / 100) * 100
        elif self.xp >= 150:
            progress = ((self.xp - 150) / 150) * 100
        else:
            progress = (self.xp / 150) * 100

        return int(progress)


player = Player("Аружан", 225)

print("👤 Оқушы:", player.name)
print("⚡ XP:", player.xp)
print("📈 Келесі деңгейге прогресс:", player.get_progress(), "%")
```

```python
# 🐍 PyStar — 216-Миссия

class Player:
    def __init__(self, name, xp):
        self.name = name
        self.xp = xp

    def get_progress(self):
        if self.xp >= 500:
            return 100
        elif self.xp >= 400:
            return int((self.xp - 400) / 100 * 100)
        elif self.xp >= 300:
            return int((self.xp - 300) / 100 * 100)
        elif self.xp >= 150:
            return int((self.xp - 150) / 150 * 100)
        else:
            return int(self.xp / 150 * 100)

    def show_progress(self):
        progress = self.get_progress()

        blocks = progress // 10
        bar = "🟩" * blocks + "⬜" * (10 - blocks)

        print("👤 Оқушы:", self.name)
        print("⚡ XP:", self.xp)
        print("📈 Прогресс:", bar)
        print("📊", progress, "%")


player = Player("Аружан", 225)

player.show_progress()
```

```python
# 🐍 PyStar — 217-Миссия

class Player:
    def __init__(self, name):
        self.name = name
        self.xp = 0
        self.starcoin = 0
        self.missions_completed = 0

    def complete_mission(self, xp_bonus, starcoin_bonus):
        self.xp += xp_bonus
        self.starcoin += starcoin_bonus
        self.missions_completed += 1

    def get_progress(self):
        if self.xp >= 500:
            return 100
        elif self.xp >= 400:
            return int((self.xp - 400) / 100 * 100)
        elif self.xp >= 300:
            return int((self.xp - 300) / 100 * 100)
        elif self.xp >= 150:
            return int((self.xp - 150) / 150 * 100)
        else:
            return int(self.xp / 150 * 100)

    def show_profile(self):
        progress = self.get_progress()
        blocks = progress // 10
        bar = "🟩" * blocks + "⬜" * (10 - blocks)

        print("================================")
        print("🤖 PyStar PLAYER")
        print("================================")
        print("👤 Оқушы:", self.name)
        print("🎯 Миссия:", self.missions_completed)
        print("⚡ XP:", self.xp)
        print("🪙 StarCoin:", self.starcoin)
        print("📈 Прогресс:", bar, progress, "%")
        print("================================")


player = Player("Аружан")

player.complete_mission(50, 10)
player.complete_mission(75, 20)
player.complete_mission(100, 30)

player.show_profile()
```

```python
# 🐍 PyStar — 222-Миссия

class Player:
    def __init__(self, name):
        self.name = name
        self.xp = 0
        self.starcoin = 0
        self.missions_completed = 0

    def answer_mission(self, answer, correct_answer):
        if answer.strip().lower() == correct_answer.lower():
            self.xp += 50
            self.starcoin += 10
            self.missions_completed += 1
            print("✅ Дұрыс! +50 XP | +10 StarCoin")
            return True
        else:
            print("❌ Қате!")
            return False

    def show_profile(self):
        print("\n🤖 PyStar PROFILE")
        print("👤 Оқушы:", self.name)
        print("⚡ XP:", self.xp)
        print("🪙 StarCoin:", self.starcoin)
        print("🎯 Дұрыс орындалған:", self.missions_completed)


questions = [
    {
        "question": "Python-да экранға мәтін шығару командасы?",
        "answer": "print"
    },
    {
        "question": "Python-да тізім құру үшін қандай жақша қолданылады?",
        "answer": "[]"
    },
    {
        "question": "Python-да шарт тексеру үшін қандай сөз қолданылады?",
        "answer": "if"
    }
]

player = Player("Аружан")

print("🎮 PyStar — 222-Миссия")
print("3 Python тапсырмасын орындаңыз!\n")

for number, question in enumerate(questions, start=1):
    print(f"🎯 {number}-тапсырма")
    print(question["question"])

    answer = input("✏️ Жауабыңыз: ")

    player.answer_mission(answer, question["answer"])
    print()

player.show_profile()
```

```python
# 🐍 PyStar — 223-Миссия

questions = [
    {
        "question": "Python-да экранға мәтін шығару командасы?",
        "answer": "print"
    },
    {
        "question": "Python-да шарт тексеру үшін қандай сөз қолданылады?",
        "answer": "if"
    },
    {
        "question": "Python-да циклдің бір түрі?",
        "answer": "for"
    },
    {
        "question": "Python-да тізім қандай жақшамен жазылады?",
        "answer": "[]"
    }
]

correct = 0
wrong = 0

print("🎮 PyStar — 223-Миссия")
print("4 Python тапсырмасы\n")

for number, question in enumerate(questions, start=1):
    print(f"🎯 {number}-тапсырма")
    print(question["question"])

    answer = input("✏️ Жауабыңыз: ")

    if answer.strip().lower() == question["answer"].lower():
        print("✅ Дұрыс!")
        correct += 1
    else:
        print("❌ Қате!")
        wrong += 1

    print()

total = len(questions)
percentage = int((correct / total) * 100)
xp = correct * 50

print("================================")
print("🏆 ТЕСТ НӘТИЖЕСІ")
print("================================")
print("✅ Дұрыс:", correct)
print("❌ Қате:", wrong)
print("📊 Нәтиже:", percentage, "%")
print("⚡ XP:", xp)
print("================================")
```

```python
# 🐍 PyStar — 236-Миссия
# ⭐ Толық оқушы профилі

def get_rewards(percentage):
    if percentage >= 90:
        return 100, 30
    elif percentage >= 75:
        return 75, 20
    elif percentage >= 50:
        return 50, 10
    else:
        return 25, 0


def get_level(percentage):
    if percentage >= 90:
        return "🔥 MASTER"
    elif percentage >= 75:
        return "🚀 PRO"
    elif percentage >= 50:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


def get_achievements(percentage):
    achievements = []

    if percentage >= 50:
        achievements.append("🥉 Алғашқы қадам")

    if percentage >= 75:
        achievements.append("🥈 Жақсы нәтиже")

    if percentage >= 90:
        achievements.append("🥇 Нағыз шебер")

    if percentage == 100:
        achievements.append("👑 PERFECT SCORE")

    return achievements


name = input("👤 Оқушының аты: ")
percentage = int(input("📊 Нәтиже (0-100): "))

xp, base_starcoin = get_rewards(percentage)
level = get_level(percentage)

achievements = get_achievements(percentage)
achievement_bonus = len(achievements) * 10

total_starcoin = base_starcoin + achievement_bonus

profile = {
    "name": name,
    "percentage": percentage,
    "xp": xp,
    "level": level,
    "achievements": achievements,
    "starcoin": total_starcoin
}

print("\n" + "=" * 45)
print("⭐ PYSTAR — ТОЛЫҚ ОҚУШЫ ПРОФИЛІ")
print("=" * 45)

print("👤 Аты:", profile["name"])
print("📊 Нәтиже:", profile["percentage"], "%")
print("⚡ XP:", profile["xp"])
print("🏆 Деңгей:", profile["level"])
print("🪙 StarCoin:", profile["starcoin"])

print("\n🏅 ЖЕТІСТІКТЕР:")

if profile["achievements"]:
    for achievement in profile["achievements"]:
        print(achievement)
else:
    print("🌱 Әзірге жетістік жоқ.")

print("=" * 45)
```

```python
# 🐍 PyStar — 237-Миссия
# 🏆 Толық профильдер және XP рейтингі

def get_rewards(percentage):
    if percentage >= 90:
        return 100, 30
    elif percentage >= 75:
        return 75, 20
    elif percentage >= 50:
        return 50, 10
    else:
        return 25, 0


def get_level(percentage):
    if percentage >= 90:
        return "🔥 MASTER"
    elif percentage >= 75:
        return "🚀 PRO"
    elif percentage >= 50:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


def get_achievements(percentage):
    achievements = []

    if percentage >= 50:
        achievements.append("🥉 Алғашқы қадам")

    if percentage >= 75:
        achievements.append("🥈 Жақсы нәтиже")

    if percentage >= 90:
        achievements.append("🥇 Нағыз шебер")

    if percentage == 100:
        achievements.append("👑 PERFECT SCORE")

    return achievements


students = [
    {"name": "Аружан", "percentage": 95},
    {"name": "Мадина", "percentage": 82},
    {"name": "Дана", "percentage": 68},
    {"name": "Нұрбек", "percentage": 91},
    {"name": "Айбек", "percentage": 45}
]


for student in students:

    xp, base_starcoin = get_rewards(student["percentage"])

    achievements = get_achievements(student["percentage"])

    achievement_bonus = len(achievements) * 10

    student["xp"] = xp
    student["level"] = get_level(student["percentage"])
    student["achievements"] = achievements
    student["starcoin"] = base_starcoin + achievement_bonus


students.sort(
    key=lambda student: student["xp"],
    reverse=True
)


print("🏆 PYSTAR — XP РЕЙТИНГ")
print("=" * 50)

for rank, student in enumerate(students, start=1):

    if rank == 1:
        medal = "🥇"
    elif rank == 2:
        medal = "🥈"
    elif rank == 3:
        medal = "🥉"
    else:
        medal = "⭐"

    print(
        medal,
        rank,
        student["name"],
        "| XP:", student["xp"],
        "| 🪙", student["starcoin"],
        "|", student["level"]
    )
```

```python
# 🐍 PyStar — 243-Миссия
# 📊 XP прогресс жолағы

def get_progress(xp):
    if xp >= 500:
        return 100

    if xp >= 400:
        progress = (xp - 400) / 100 * 100
    elif xp >= 300:
        progress = (xp - 300) / 100 * 100
    elif xp >= 150:
        progress = (xp - 150) / 150 * 100
    else:
        progress = xp / 150 * 100

    return int(progress)


xp = int(input("⚡ XP енгізіңіз: "))

progress = get_progress(xp)

filled = int(progress / 10)
bar = "🟩" * filled + "⬜" * (10 - filled)

print("\n⭐ PYSTAR — XP ПРОГРЕСС")
print("=" * 35)
print(bar)
print("📊 Прогресс:", progress, "%")
```

```python

# 🐍 PyStar — 251-Миссия
# 🎯 Миссия орындау → XP + StarCoin

def complete_mission(score):
    if score >= 90:
        xp = 100
        starcoin = 30
    elif score >= 75:
        xp = 75
        starcoin = 20
    elif score >= 50:
        xp = 50
        starcoin = 10
    else:
        xp = 25
        starcoin = 0

    return xp, starcoin


score = int(input("📊 Миссия нәтижесі (0-100): "))

xp, starcoin = complete_mission(score)

print("\n🎯 МИССИЯ АЯҚТАЛДЫ!")
print("📊 Нәтиже:", score, "%")
print("⚡ Алынған XP:", xp)
print("🪙 Алынған StarCoin:", starcoin)
```

```python
# 🐍 PyStar — 252-миссия
# 🎯 Бірнеше миссия орындау

def complete_mission(score):
    if score >= 90:
        return 100, 30
    elif score >= 75:
        return 75, 20
    elif score >= 50:
        return 50, 10
    else:
        return 25, 0


xp = 0
starcoin = 0

for i in range(1, 4):
    print("\n🎯", i, "-миссия")
    score = int(input("📊 Нәтиже (0-100): "))

    mission_xp, mission_starcoin = complete_mission(score)

    xp += mission_xp
    starcoin += mission_starcoin

print("\n" + "=" * 40)
print("⭐ PYSTAR — МИССИЯ ҚОРЫТЫНДЫСЫ")
print("=" * 40)
print("⚡ Жалпы XP:", xp)
print("🪙 Жалпы StarCoin:", starcoin)
print("=" * 40)
```

```python
# 🐍 PyStar — 253-миссия
# 🎯 Миссия саны + XP + StarCoin

def complete_mission(score):
    if score >= 90:
        return 100, 30
    elif score >= 75:
        return 75, 20
    elif score >= 50:
        return 50, 10
    else:
        return 25, 0


missions_completed = 0
xp = 0
starcoin = 0

for i in range(1, 4):
    print("\n🎯", i, "-миссия")
    score = int(input("📊 Нәтиже (0-100): "))

    mission_xp, mission_starcoin = complete_mission(score)

    missions_completed += 1
    xp += mission_xp
    starcoin += mission_starcoin

print("\n" + "=" * 40)
print("⭐ PYSTAR — ОҚУШЫ НӘТИЖЕСІ")
print("=" * 40)
print("🎯 Орындалған миссия:", missions_completed)
print("⚡ Жалпы XP:", xp)
print("🪙 Жалпы StarCoin:", starcoin)
print("=" * 40)
```

```python
# 🐍 PyStar — 254-миссия
# 🏆 XP → Деңгей

def get_level(xp):
    if xp >= 500:
        return "👑 LEGEND"
    elif xp >= 400:
        return "🔥 MASTER"
    elif xp >= 300:
        return "🚀 PRO"
    elif xp >= 150:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


xp = int(input("⚡ Оқушының XP саны: "))

level = get_level(xp)

print("\n" + "=" * 40)
print("⭐ PYSTAR — ДЕҢГЕЙ НӘТИЖЕСІ")
print("=" * 40)
print("⚡ XP:", xp)
print("🏆 Деңгей:", level)
print("=" * 40)
```

```python
# 🐍 PyStar — 255-миссия
# ⚡ Келесі деңгейге қалған XP

def get_level_info(xp):
    if xp >= 500:
        return "👑 LEGEND", 0
    elif xp >= 400:
        return "🔥 MASTER", 500 - xp
    elif xp >= 300:
        return "🚀 PRO", 400 - xp
    elif xp >= 150:
        return "⭐ ADVANCED", 300 - xp
    else:
        return "🌱 BEGINNER", 150 - xp


xp = int(input("⚡ XP енгізіңіз: "))

level, needed_xp = get_level_info(xp)

print("\n" + "=" * 40)
print("⭐ PYSTAR — XP ПРОГРЕСС")
print("=" * 40)
print("⚡ Қазіргі XP:", xp)
print("🏆 Деңгей:", level)

if needed_xp == 0:
    print("👑 Ең жоғары деңгейге жеттіңіз!")
else:
    print("🎯 Келесі деңгейге қалған XP:", needed_xp)

print("=" * 40)
```

```python
# 🐍 PyStar — 256-миссия
# 📊 XP прогресс жолағы

def get_progress(xp):
    if xp >= 500:
        return 100
    elif xp >= 400:
        return (xp - 400) / 100 * 100
    elif xp >= 300:
        return (xp - 300) / 100 * 100
    elif xp >= 150:
        return (xp - 150) / 150 * 100
    else:
        return xp / 150 * 100


xp = int(input("⚡ XP енгізіңіз: "))

progress = int(get_progress(xp))

filled = int(progress / 10)
bar = "🟩" * filled + "⬜" * (10 - filled)

print("\n" + "=" * 40)
print("⭐ PYSTAR — XP ПРОГРЕССІ")
print("=" * 40)
print("⚡ XP:", xp)
print("📊 Прогресс:", progress, "%")
print(bar)
print("=" * 40)
```

```python
# 🐍 PyStar — 257-миссия
# ⭐ Оқушының толық XP прогресі

def get_level(xp):
    if xp >= 500:
        return "👑 LEGEND"
    elif xp >= 400:
        return "🔥 MASTER"
    elif xp >= 300:
        return "🚀 PRO"
    elif xp >= 150:
        return "⭐ ADVANCED"
    else:
        return "🌱 BEGINNER"


def get_next_xp(xp):
    if xp >= 500:
        return 0
    elif xp >= 400:
        return 500 - xp
    elif xp >= 300:
        return 400 - xp
    elif xp >= 150:
        return 300 - xp
    else:
        return 150 - xp


def get_progress(xp):
    if xp >= 500:
        return 100
    elif xp >= 400:
        return (xp - 400) / 100 * 100
    elif xp >= 300:
        return (xp - 300) / 100 * 100
    elif xp >= 150:
        return (xp - 150) / 150 * 100
    else:
        return xp / 150 * 100


xp = int(input("⚡ XP енгізіңіз: "))

level = get_level(xp)
next_xp = get_next_xp(xp)
progress = int(get_progress(xp))

filled = int(progress / 10)
bar = "🟩" * filled + "⬜" * (10 - filled)

print("\n" + "=" * 45)
print("⭐ PYSTAR — ОҚУШЫ XP ПРОФИЛІ")
print("=" * 45)
print("⚡ XP:", xp)
print("🏆 Деңгей:", level)
print("📊 Прогресс:", progress, "%")
print(bar)

if next_xp == 0:
    print("👑 Барлық деңгейлерді аяқтадыңыз!")
else:
    print("🎯 Келесі деңгейге қалған XP:", next_xp)

print("=" * 45)
```

```python
# 🐍 PyStar — 273-миссия
# 🥈 Екінші ең үлкен санды табу

numbers = [12, 45, 7, 89, 34, 56, 23]

largest = None
second_largest = None

for number in numbers:
    if largest is None or number > largest:
        second_largest = largest
        largest = number
    elif number != largest and (second_largest is None or number > second_largest):
        second_largest = number

print("Ең үлкен сан:", largest)
print("🥈 Екінші ең үлкен сан:", second_largest)
```

```python
# 🌟 PyStar — Біртұтас ойынның алғашқы нұсқасы

print("\n" + "=" * 60)
print("🌟 P Y S T A R 🌟")
print("🐍 Python үйрен — жұлдыз бол!")
print("=" * 60)

# 👤 Оқушы
name = input("\n👤 Оқушының аты: ")
grade = input("🎓 Сыныбы (7/8/9/10): ")

# 🎮 Бастапқы көрсеткіштер
xp = 0
starcoin = 0

while True:

    print("\n" + "-" * 50)
    print(f"👤 {name} | 🎓 {grade}")
    print(f"⚡ XP: {xp} | 🪙 StarCoin: {starcoin}")
    print("-" * 50)

    print("\n🎮 PYSTAR МӘЗІРІ")
    print("1️⃣ 📚 Оқу әлемі")
    print("2️⃣ 🎯 Миссия бастау")
    print("3️⃣ 🏆 Деңгейім")
    print("4️⃣ 🎖️ Жетістіктерім")
    print("5️⃣ 👤 Профилім")
    print("0️⃣ 🚪 Шығу")

    choice = input("\n👉 Таңдаңыз: ")

    if choice == "1":
        print("\n📚 ОҚУ ӘЛЕМІ")
        print("🐍 Python")
        print("🔀 Шарттар")
        print("🔁 Циклдер")
        print("📋 Тізімдер")
        print("🧠 Алгоритмдер")

    elif choice == "2":
        print("\n🎯 МИССИЯ")
        print("⭐ Миссия жүйесі дайындалып жатыр.")
        print("📚 Мұнда кейін 273+ нақты Python миссиясы қосылады.")

    elif choice == "3":
        if xp < 100:
            level = "STARTER"
        elif xp < 300:
            level = "BEGINNER"
        elif xp < 600:
            level = "ADVANCED"
        elif xp < 1000:
            level = "EXPERT"
        else:
            level = "MASTER"

        print("\n🏆 ДЕҢГЕЙІҢІЗ:", level)
        print("⚡ XP:", xp)

    elif choice == "4":
        print("\n🎖️ ЖЕТІСТІКТЕР")

        if xp >= 100:
            print("✅ 🐍 Python бастаушысы")
        if xp >= 300:
            print("✅ 🔥 Код шебері")
        if xp >= 600:
            print("✅ 🚀 Python зерттеушісі")
        if xp >= 1000:
            print("✅ 👑 PyStar MASTER")

        if xp < 100:
            print("⭐ Бірінші жетістікке жету үшін 100 XP жинаңыз.")

    elif choice == "5":
        print("\n👤 МЕНІҢ ПРОФИЛІМ")
        print("👤 Аты:", name)
        print("🎓 Сыныбы:", grade)
        print("⚡ XP:", xp)
        print("🪙 StarCoin:", starcoin)

    elif choice == "0":
        print("\n👋 PyStar ойынынан шықтыңыз!")
        print("🌟 Қайта кездескенше, Python жұлдызы!")
        break

    else:
        print("\n❗ Қате таңдау. 0–5 аралығындағы санды енгізіңіз.")
```

```python
# 🌟 PYSTAR — БІРТҰТАС ОҚУ ОЙЫНЫ
# 📚 Оқу әлемі
# 🎯 Миссиялар
# 🧠 Жауапты тексеру + түсіндіру
# ⚡ XP + 🪙 StarCoin
# 🏆 Деңгей
# 🎖️ Жетістіктер
# 👤 Оқушы профилі
# 💾 Ойынның негізгі құрылымы


# =========================================================
# 🌟 PYSTAR
# =========================================================

print("\n" + "=" * 60)
print("🌟  P Y S T A R  🌟")
print("🐍 Python үйрен — жұлдыз бол!")
print("=" * 60)


# =========================================================
# 📚 ОҚУ ӘЛЕМІ
# =========================================================

def оқу_әлемі():

    print("\n" + "=" * 55)
    print("📚 PYSTAR — ОҚУ ӘЛЕМІ")
    print("=" * 55)

    world_grade = input("\n🎓 Сыныбыңызды таңдаңыз (7/8/9/10): ")

    if world_grade == "7":

        print("\n🐍 7-СЫНЫП")
        print("1 — Python негіздері")
        print("2 — Айнымалылар және деректер типтері")
        print("3 — if / else шарттары")
        print("4 — for және while циклдері")
        print("5 — Тізімдер")

    elif world_grade == "8":

        print("\n🐍 8-СЫНЫП")
        print("1 — Күрделі шарттар")
        print("2 — Циклдерді басқару")
        print("3 — Тізімдермен жұмыс")
        print("4 — Жолдар (str)")
        print("5 — Функциялар")

    elif world_grade == "9":

        print("\n🐍 9-СЫНЫП")
        print("1 — Функциялар")
        print("2 — Сөздіктер (dict)")
        print("3 — Файлдармен жұмыс")
        print("4 — Алгоритмдер")
        print("5 — Қателерді өңдеу")

    elif world_grade == "10":

        print("\n🐍 10-СЫНЫП")
        print("1 — Күрделі алгоритмдер")
        print("2 — Деректер құрылымдары")
        print("3 — Функциялар және модульдер")
        print("4 — Алгоритмдік есептер")
        print("5 — Python жобалары")

    else:

        print("\n❗ 7, 8, 9 немесе 10 сыныпты таңдаңыз.")


# =========================================================
# 👤 ОҚУШЫ ПРОФИЛІ
# =========================================================

name = input("\n👤 Оқушының аты: ")
grade = input("🎓 Сыныбы (7/8/9/10): ")


# =========================================================
# 🎮 ОЙЫН КӨРСЕТКІШТЕРІ
# =========================================================

xp = 0
starcoin = 0

# Орындалған миссиялардың нөмірлерін сақтаймыз
completed_missions = []


# =========================================================
# 🏆 ДЕҢГЕЙДІ АНЫҚТАУ
# =========================================================

def деңгей_анықтау(xp):

    if xp < 100:
        return "STARTER"

    elif xp < 300:
        return "BEGINNER"

    elif xp < 600:
        return "ADVANCED"

    elif xp < 1000:
        return "EXPERT"

    else:
        return "MASTER"


# =========================================================
# 🎖️ ЖЕТІСТІКТЕРДІ КӨРСЕТУ
# =========================================================

def жетістіктерді_көрсету(xp):

    print("\n" + "=" * 50)
    print("🎖️ ЖЕТІСТІКТЕРІМ")
    print("=" * 50)

    if xp >= 100:
        print("✅ 🐍 Python бастаушысы")

    if xp >= 300:
        print("✅ 🔥 Код шебері")

    if xp >= 600:
        print("✅ 🚀 Python зерттеушісі")

    if xp >= 1000:
        print("✅ 👑 PyStar MASTER")

    if xp < 100:
        print("⭐ Алғашқы жетістікке 100 XP керек.")


# =========================================================
# 🎯 МИССИЯНЫ ОРЫНДАУ ФУНКЦИЯСЫ
# =========================================================

def миссия_263():

    global xp, starcoin

    print("\n" + "=" * 55)
    print("🐍 МИССИЯ №263")
    print("=" * 55)

    print("📚 Тақырып: Тізімдегі ең үлкен санды табу")
    print("🎓 Деңгей: 7–10 сынып")

    print("\n🧠 Тапсырма:")
    print("Берілген тізімдегі ең үлкен санды табыңыз.")
    print("❗ max() функциясын қолдануға болмайды.")

    print("\n📋 Берілген тізім:")
    print("[12, 45, 7, 89, 34, 56]")

    while True:

        answer = input("\n👉 Ең үлкен сан неше? ")

        if answer == "89":

            print("\n✅ ДҰРЫС!")
            print("🎉 Миссия орындалды!")

            xp += 30
            starcoin += 10

            print("⚡ +30 XP")
            print("🪙 +10 StarCoin")

            print("\n💡 Түсіндіру:")
            print("Тізімдегі сандарды салыстырғанда ең үлкен сан — 89.")

            return True

        else:

            print("\n❌ Қате жауап.")
            print("💡 Кеңес: барлық сандарды бір-бірімен салыстырып көріңіз.")

            retry = input("\n🔁 Қайта жауап бересіз бе? (иә/жоқ): ")

            if retry.lower() != "иә":
                print("➡️ Миссия кейін қайта орындалуы мүмкін.")
                return False


# =========================================================
# 🥈 МИССИЯ №273
# =========================================================

def миссия_273():

    global xp, starcoin

    print("\n" + "=" * 55)
    print("🐍 МИССИЯ №273")
    print("=" * 55)

    print("🥈 Тақырып: Екінші ең үлкен санды табу")
    print("🎓 Деңгей: 7–10 сынып")

    print("\n🧠 Тапсырма:")
    print("Берілген тізімдегі екінші ең үлкен санды табыңыз.")

    print("\n📋 Берілген тізім:")
    print("[12, 45, 7, 89, 34, 56, 23]")

    while True:

        answer = input("\n👉 Екінші ең үлкен сан неше? ")

        if answer == "56":

            print("\n✅ ДҰРЫС!")
            print("🎉 Миссия орындалды!")

            xp += 30
            starcoin += 10

            print("⚡ +30 XP")
            print("🪙 +10 StarCoin")

            print("\n💡 Түсіндіру:")
            print("Ең үлкен сан — 89.")
            print("Одан кейінгі екінші үлкен сан — 56.")

            return True

        else:

            print("\n❌ Қате жауап.")
            print("💡 Кеңес: алдымен ең үлкен санды тап,")
            print("содан кейін келесі үлкен санды анықта.")

            retry = input("\n🔁 Қайта жауап бересіз бе? (иә/жоқ): ")

            if retry.lower() != "иә":
                print("➡️ Миссия кейін қайта орындалуы мүмкін.")
                return False


# =========================================================
# 🎯 МИССИЯЛАР МӘЗІРІ
# =========================================================

def миссиялар():

    global completed_missions

    print("\n" + "=" * 55)
    print("🎯 PYSTAR — МИССИЯЛАР")
    print("=" * 55)

    print("\n263 — 📚 Ең үлкен санды табу")
    print("273 — 🥈 Екінші ең үлкен санды табу")

    print("\n0 — ↩️ Артқа")

    mission = input("\n👉 Миссия нөмірін таңдаңыз: ")

    if mission == "263":

        if 263 in completed_missions:

            print("\nℹ️ Бұл миссия бұрын орындалған.")

        else:

            нәтиже = миссия_263()

            if нәтиже:
                completed_missions.append(263)

    elif mission == "273":

        if 273 in completed_missions:

            print("\nℹ️ Бұл миссия бұрын орындалған.")

        else:

            нәтиже = миссия_273()

            if нәтиже:
                completed_missions.append(273)

    elif mission == "0":

        print("\n↩️ Негізгі мәзірге қайттыңыз.")

    else:

        print("\n❗ Мұндай миссия әзірге жоқ.")


# =========================================================
# 🏆 ДЕҢГЕЙІМ
# =========================================================

def деңгейім():

    level = деңгей_анықтау(xp)

    print("\n" + "=" * 50)
    print("🏆 МЕНІҢ ДЕҢГЕЙІМ")
    print("=" * 50)

    print("👤 Оқушы:", name)
    print("⚡ XP:", xp)
    print("🏆 Level:", level)

    print("\n📈 Келесі деңгейлер:")

    if level == "STARTER":
        print("➡️ BEGINNER — 100 XP")

    elif level == "BEGINNER":
        print("➡️ ADVANCED — 300 XP")

    elif level == "ADVANCED":
        print("➡️ EXPERT — 600 XP")

    elif level == "EXPERT":
        print("➡️ MASTER — 1000 XP")

    else:
        print("👑 Сіз MASTER деңгейіндесіз!")


# =========================================================
# 👤 ПРОФИЛЬ
# =========================================================

def профиль():

    level = деңгей_анықтау(xp)

    print("\n" + "=" * 50)
    print("👤 МЕНІҢ ПРОФИЛІМ")
    print("=" * 50)

    print("👤 Оқушы:", name)
    print("🎓 Сынып:", grade)
    print("⚡ XP:", xp)
    print("🪙 StarCoin:", starcoin)
    print("🏆 Деңгей:", level)

    print("🎯 Орындалған миссия саны:",
          len(completed_missions))

    if completed_missions:

        print("📋 Миссиялар:",
              completed_missions)

    else:

        print("📋 Әзірге миссия орындалмаған.")


# =========================================================
# 🎮 НЕГІЗГІ ОЙЫН
# =========================================================

while True:

    print("\n" + "-" * 55)

    print(f"👤 {name} | 🎓 {grade}-сынып")
    print(f"⚡ XP: {xp} | 🪙 StarCoin: {starcoin}")
    print(f"🎯 Орындалған миссия: {len(completed_missions)}")
    print(f"🏆 Деңгей: {деңгей_анықтау(xp)}")

    print("-" * 55)

    print("\n🎮 PYSTAR МӘЗІРІ")

    print("1️⃣ 📚 Оқу әлемі")
    print("2️⃣ 🎯 Миссия бастау")
    print("3️⃣ 🏆 Деңгейім")
    print("4️⃣ 🎖️ Жетістіктерім")
    print("5️⃣ 👤 Профилім")
    print("0️⃣ 🚪 Шығу")

    choice = input("\n👉 Таңдаңыз: ")


    # -----------------------------------------------------
    # 📚 ОҚУ ӘЛЕМІ
    # -----------------------------------------------------

    if choice == "1":

        оқу_әлемі()


    # -----------------------------------------------------
    # 🎯 МИССИЯ
    # -----------------------------------------------------

    elif choice == "2":

        миссиялар()


    # -----------------------------------------------------
    # 🏆 ДЕҢГЕЙ
    # -----------------------------------------------------

    elif choice == "3":

        деңгейім()


    # -----------------------------------------------------
    # 🎖️ ЖЕТІСТІК
    # -----------------------------------------------------

    elif choice == "4":

        жетістіктерді_көрсету(xp)


    # -----------------------------------------------------
    # 👤 ПРОФИЛЬ
    # -----------------------------------------------------

    elif choice == "5":

        профиль()


    # -----------------------------------------------------
    # 🚪 ШЫҒУ
    # -----------------------------------------------------

    elif choice == "0":

        print("\n👋 PyStar ойынынан шықтыңыз!")
        print("🌟 Қайта кездескенше, Python жұлдызы!")

        break


    # -----------------------------------------------------
    # ❗ ҚАТЕ ТАҢДАУ
    # -----------------------------------------------------

    else:

        print("\n❗ Қате таңдау.")
        print("0–5 аралығындағы санды енгізіңіз.")
```
